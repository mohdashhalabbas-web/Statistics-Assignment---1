Q4: Median in Real Life (Medium)
A car dealer notes the prices of used cars: [$8,000, $9,500, $10,200, $11,000, $50,000].
Why is the median a better measure than the mean in this case? Calculate the median.
     - Median Calculation:
There are 5 values, so the median is the middle (3rd) value.
Median = $10,200

Why is the median better than the mean?
The price $50,000 is an outlier (much higher than the other car prices). This unusually high value pulls the mean upward, making it seem that the typical car price is much higher than it actually is.

The median is not affected by this outlier, so it gives a more accurate representation of the typical used car price.

Answer: The median is $10,200, and it is a better measure than the mean because it is not influenced by the unusually expensive $50,000 car.

A student times how long it takes to finish a puzzle each day: [25, 30, 27, 35, 40].
What does the range tell us about the variation in the student’s puzzle-solving time?

  - Puzzle-solving times (in minutes): 25, 30, 27, 35, 40

Range = Highest value − Lowest value

Range=40−25=15 minutes

Answer: The range is 15 minutes.

Q6: Range in Action (Medium)
A farmer records the weekly weight of harvested apples (kg): [100, 105, 98, 110, 120].
Find the range. How can this help the farmer in planning his packaging?

  - Weekly apple harvest weights (kg): 100, 105, 98, 110, 120

Range = Highest value − Lowest value

Range=120−98=22 kg

Answer: The range is 22 kg.
How does this help the farmer?
The range shows that the weekly harvest varies by 22 kg. This helps the farmer:

Estimate the variation in harvest size each week.
Prepare enough packaging materials for weeks with larger harvests.
Plan storage and transportation more efficiently.
Avoid shortages or excess packaging by considering possible changes in harvest weight.

Q7: Variance for Decision-Making (Medium)
Two delivery companies track delivery delays (in minutes).

Company A: variance = 6

Company B: variance = 15
 Which company is more consistent, and why?

  - Answer: Company A is more consistent because it has a lower variance.

Company A: Variance = 6
Company B: Variance = 15

Variance measures how much the delivery delays vary from the average delay. A lower variance means the delivery times are more consistent and have less fluctuation.

Conclusion: Since 6 < 15, Company A has more reliable and consistent delivery times, while Company B experiences greater variation in delivery delays.

Q8: Standard Deviation in Context (Hard)
A finance student compares the daily price fluctuations of two cryptocurrencies.

Coin A: standard deviation = $30

Coin B: standard deviation = $120
 Which coin is riskier to invest in, and why?

   - Answer: Coin B is riskier to invest in because it has a higher standard deviation.

Coin A: Standard deviation = $30
Coin B: Standard deviation = $120

The standard deviation measures how much the daily prices fluctuate around the average price.

A lower standard deviation ($30) means the price is relatively stable with smaller daily changes.
A higher standard deviation ($120) means the price changes much more from day to day, making it more volatile.

Conclusion: Coin B is riskier because its price fluctuates much more than Coin A's. While higher volatility can offer the potential for greater returns, it also increases the risk of larger losses.
